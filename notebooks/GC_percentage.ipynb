```python
import numpy as np
import pandas as pd


dna = "ATCGATCGATCGCGTAGCTAGCCTAGCTAGCTAGCTACTATCGCGATATATCTACTATATAGCTAGCTAGCTAGCTACTACTG"

print(dna)
print(len(dna))

nucleotide_count = pd.DataFrame({"nucleotide": ["A", "T", "C", "G"], "count": [dna.count("A"), dna.count("T"), dna.count("C"), dna.count("G")]}
                                )

print(nucleotide_count)
```

```python
nucleotide_count["count"]

G = nucleotide_count.loc[nucleotide_count["nucleotide"] == "G", "count"].iloc[0]
C = nucleotide_count.loc[nucleotide_count["nucleotide"] == "C", "count"].iloc[0]

GC = G + C

GC_percent = GC/nucleotide_count["count"].sum() * 100

print(GC_percent)
```
